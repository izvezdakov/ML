## ex. 1

$$  
y = x^Tx,  \quad x \in \mathbb{R}^N 
$$

$$
\frac{dy}{dx} = 2x
$$

**Вывод.** Распишем $y$ через сумму:
$$
y = x^T x = \sum_{i=1}^{N} x_i^2 .
$$
Тогда
$$
\frac{\partial y}{\partial x_i} = 2x_i \quad\Longrightarrow\quad \frac{dy}{dx} = 2x .
$$


## ex. 2

$$
y = \operatorname{tr}(AB)
$$

$$
\frac{dy}{dA} = B^T
$$

**Вывод.** След — сумма диагональных элементов произведения:
$$
y = \operatorname{tr}(AB) = \sum_{i=1}^{N}\sum_{j=1}^{N} A_{ij}B_{ji}.
$$
Поэтому
$$
\frac{\partial y}{\partial A_{ij}} = B_{ji} \quad\Longrightarrow\quad \frac{dy}{dA} = B^T .
$$


## ex. 3

$$  
y = x^TAc , \quad A\in \mathbb{R}^{N \times N}, x\in \mathbb{R}^{N}, c\in \mathbb{R}^{N} 
$$

$$
\frac{dy}{dx} = Ac
$$

**Вывод.** $y=x^T(Ac)$ — линейная форма по $x$ с вектором коэффициентов $Ac$:
$$
y=\sum_{i=1}^{N} x_i (Ac)_i \quad\Longrightarrow\quad \frac{\partial y}{\partial x_i}=(Ac)_i,
$$
то есть $\frac{dy}{dx}=Ac$.

$$
\frac{dy}{dA} = xc^T
$$

**Вывод.** $y$ — скаляр, поэтому его можно записать под следом и воспользоваться цикличностью:
$$
y = x^TAc = \operatorname{tr}(x^TAc) = \operatorname{tr}(cx^TA).
$$
Применяем результат ex.2 к $B = cx^T$:
$$
\frac{dy}{dA} = B^T = (cx^T)^T = xc^T .
$$
Размерности: $x\in\mathbb{R}^{N}$, $c^T\in\mathbb{R}^{1\times N}$, значит $xc^T\in\mathbb{R}^{N\times N}$ — как и $A$.

## ex. 4

Classic matrix factorization example. Given matrix $X$ you need to find $A$, $S$ to approximate $X$. This can be done by simple gradient descent iteratively alternating $A$ and $S$ updates.
$$
J = || X - AS ||_F^2  , \quad A\in \mathbb{R}^{N \times R} , \quad S\in \mathbb{R}^{R \times M}
$$
$$
\frac{dJ}{dS} = ? 
$$

You may use one of the following approaches:

#### First approach
Using ex.2 and the fact:
$$
|| X ||_F^2 = tr(XX^T) 
$$ 
it is easy to derive gradients (you can find it in one of the refs). 

#### Second approach
You can use *slightly different techniques* if they suits you. Take a look at this derivation:
<img src="grad.png">
(excerpt from [Handbook of blind source separation, Jutten, page 517](https://books.google.ru/books?id=PTbj03bYH6kC&printsec=frontcover&dq=Handbook+of+Blind+Source+Separation&hl=en&sa=X&ved=0ahUKEwi-q_apiJDLAhULvXIKHVXJDWcQ6AEIHDAA#v=onepage&q=Handbook%20of%20Blind%20Source%20Separation&f=false), open for better picture).

#### Third approach
And finally we can use chain rule!
let $F = AS$.

**Find**
$$
\frac{dJ}{dF} = 2(F-X) = -2(X-AS)
$$
and
$$
\frac{dF}{dS} = I_M \otimes A
$$
(the shape should be $NM \times RM$; это якобиан линейного отображения $\operatorname{vec}(F)=(I_M\otimes A)\operatorname{vec}(S)$ при векторизации по столбцам).

Now it is easy do get desired gradients:
$$
\operatorname{vec}\!\left(\frac{dJ}{dS}\right)=\left(I_M\otimes A\right)^T\operatorname{vec}\!\left(\frac{dJ}{dF}\right),
\qquad\text{то есть}\qquad
\frac{dJ}{dS} = -2A^T(X-AS).
$$

#### Проверка: first approach

$$
J=\|X-AS\|_F^2=\operatorname{tr}\!\big((X-AS)^T(X-AS)\big).
$$
Раскрываем и группируем перекрёстные слагаемые:
$$
J=\operatorname{tr}(X^TX)-2\operatorname{tr}(X^TAS)+\operatorname{tr}(S^TA^TAS).
$$
Дифференцируем по $S$, используя $\frac{\partial}{\partial S}\operatorname{tr}(PS)=P^T$ и симметричность $A^TA$:
$$
\frac{dJ}{dS}=-2A^TX+2A^TAS=-2A^T(X-AS).
$$

### Question 2

We can also use other distance metrics such as L1 distance.
For pixel values $p_{ij}^{(k)}$ at location $(i,j)$ of some image $I_k$, 

the mean $\mu$ across all pixels over all images is $$\mu=\frac{1}{nhw}\sum_{k=1}^n\sum_{i=1}^{h}\sum_{j=1}^{w}p_{ij}^{(k)}$$
And the pixel-wise mean $\mu_{ij}$ across all images is 
$$\mu_{ij}=\frac{1}{n}\sum_{k=1}^np_{ij}^{(k)}.$$
The general standard deviation $\sigma$ and pixel-wise standard deviation $\sigma_{ij}$ is defined similarly.

Which of the following preprocessing steps will not change the performance of a Nearest Neighbor classifier that uses L1 distance? Select all that apply.
1. Subtracting the mean $\mu$ ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu$.)
2. Subtracting the per pixel mean $\mu_{ij}$  ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu_{ij}$.)
3. Subtracting the mean $\mu$ and dividing by the standard deviation $\sigma$.
4. Subtracting the pixel-wise mean $\mu_{ij}$ and dividing by the pixel-wise standard deviation $\sigma_{ij}$.
5. Rotating the coordinate axes of the data.

*Your Answer:* 1, 2, 3


*Your Explanation:*

Обозначим L1-расстояние между изображениями $X$ и $Y$: $d(X,Y)=\sum_{i,j}|X_{ij}-Y_{ij}|$.

1. Вычитание общего среднего $\mu$: $|(X_{ij}-\mu)-(Y_{ij}-\mu)|=|X_{ij}-Y_{ij}|$ — расстояния не меняются.
2. Вычитание среднего по пикселю $\mu_{ij}$: величина $\mu_{ij}$ одинакова для всех изображений, поэтому тоже сокращается в разности — расстояния не меняются.
3. Вычитание $\mu$ и деление на $\sigma$: $d' = d/\sigma$. Все расстояния умножаются на одну и ту же положительную константу, порядок соседей сохраняется, предсказания 1-NN не меняются.
4. Деление на по-пиксельное $\sigma_{ij}$ даёт разным пикселям разные веса: $d'=\sum_{i,j}\frac{1}{\sigma_{ij}}|X_{ij}-Y_{ij}|$ — порядок соседей может измениться.
5. L1-метрика не инвариантна к повороту осей (в отличие от L2): $\|Q(X-Y)\|_1 \neq \|X-Y\|_1$, где $Q$ — матрица поворота.
